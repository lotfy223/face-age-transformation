```python
import glob
import torch
import matplotlib.pyplot as plt

from PIL import Image, ImageOps
from diffusers import StableDiffusionInstructPix2PixPipeline

# نتأكد إن الـGPU شغال
if not torch.cuda.is_available():
    raise RuntimeError(
        "الـGPU مش شغال. فعّله من Runtime > Change runtime type."
    )

print("GPU:", torch.cuda.get_device_name(0))

# نختار أول صورة من الوجوه اللي ولّدناها قبل كده
image_files = sorted(
    glob.glob("/content/stylegan2-ada-pytorch/generated_faces/*.png")
)

if not image_files:
    raise FileNotFoundError(
        "مش لاقي الصور المولّدة. شغّل خلية توليد الوجوه الأصلية أولًا."
    )

image_path = image_files[0]
original = Image.open(image_path).convert("RGB")
original = ImageOps.fit(original, (512, 512))

print("بنستخدم الصورة:", image_path)
print("جاري تحميل موديل تعديل الصور... أول مرة ممكن تاخد وقت.")

pipe = StableDiffusionInstructPix2PixPipeline.from_pretrained(
    "timbrooks/instruct-pix2pix",
    torch_dtype=torch.float16,
)
pipe = pipe.to("cuda")
pipe.enable_attention_slicing()

# أوصاف الأعمار اللي عايزين نجربها
ages = [
    ("Child", "Make this person look like a child, preserving facial features."),
    ("Young adult", "Make this person look like a young adult, preserving facial features."),
    ("Older adult", "Make this person look elderly, preserving facial features."),
]

results = []

for title, prompt in ages:
    print("جاري توليد:", title)

    generator = torch.Generator("cuda").manual_seed(42)

    result = pipe(
        prompt=prompt,
        image=original,
        num_inference_steps=25,
        guidance_scale=7.5,
        image_guidance_scale=1.5,
        generator=generator,
    ).images[0]

    results.append((title, result))

# عرض الأصل والمراحل العمرية
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

axes[0].imshow(original)
axes[0].set_title("Original")
axes[0].axis("off")

for ax, (title, image) in zip(axes[1:], results):
    ax.imshow(image)
    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import re
from collections import Counter

import matplotlib.pyplot as plt
from datasets import load_dataset

dataset_id = "py97/UTKFace-Cropped"

print("جاري فتح البيانات...")

# streaming بيقرأ البيانات تدريجيًا بدل تحميل الداتا كلها مرة واحدة
dataset = load_dataset(
    dataset_id,
    split="train",
    streaming=True,
)

counts = Counter()
examples = {}
invalid = 0
total = 0

for row in dataset.take(1000):
    total += 1

    # مثال: UTKFace/82_0_2_...
    key = row.get("__key__", "")
    match = re.search(r"(?:^|/)(\d+)_", key)

    if not match:
        invalid += 1
        continue

    age = int(match.group(1))

    if not 0 <= age <= 120:
        invalid += 1
        continue

    if age <= 12:
        group = "Child (0-12)"
    elif age <= 25:
        group = "Young (13-25)"
    elif age <= 59:
        group = "Adult (26-59)"
    else:
        group = "Older (60+)"

    counts[group] += 1

    # نحتفظ بمثال واحد للعرض من كل فئة
    if group not in examples:
        examples[group] = (age, row["jpg.chip.jpg"])

print("\nعدد الصور المقروءة:", total)
print("صور لم نستطع استخراج سن صالح لها:", invalid)

groups = [
    "Child (0-12)",
    "Young (13-25)",
    "Adult (26-59)",
    "Older (60+)",
]

print("\nتوزيع أول 1000 صورة:")
for group in groups:
    print(f"{group}: {counts[group]}")

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, group in zip(axes, groups):
    ax.axis("off")

    if group in examples:
        age, image = examples[group]
        ax.imshow(image)
        ax.set_title(f"{group}\nRecorded age: {age}")
    else:
        ax.set_title(f"{group}\nNo example found")

plt.tight_layout()
plt.show()
```

```python
import os
import re
import hashlib
from collections import Counter

from datasets import load_dataset
from PIL import ImageOps

BASE = "/content/age_project_data"
LIMIT = 1000

for split in ("train", "test"):
    os.makedirs(f"{BASE}/{split}", exist_ok=True)

dataset = load_dataset(
    "py97/UTKFace-Cropped",
    split="train",
    streaming=True,
)

counts = Counter()

for index, row in enumerate(dataset.take(LIMIT), start=1):
    try:
        key = row["__key__"]
        match = re.search(r"(?:^|/)(\d+)_", key)

        if not match:
            counts["invalid"] += 1
            continue

        age = int(match.group(1))
        if not 0 <= age <= 120:
            counts["invalid"] += 1
            continue

        digest = hashlib.sha256(key.encode()).hexdigest()
        split = "test" if int(digest[:8], 16) % 10 == 0 else "train"

        image = row["jpg.chip.jpg"].convert("RGB")
        image = ImageOps.fit(image, (256, 256))

        path = f"{BASE}/{split}/{age}_{digest[:16]}.jpg"
        image.save(path, quality=95)
        counts[split] += 1

        if index % 100 == 0:
            print(f"تمت معالجة {index} صورة")

    except Exception as error:
        counts["errors"] += 1
        if counts["errors"] <= 3:
            print(f"مشكلة في الصورة {index}: {error}")

print("\nالنتيجة:")
print("تدريب:", counts["train"])
print("اختبار:", counts["test"])
print("أعمار غير صالحة:", counts["invalid"])
print("أخطاء:", counts["errors"])
```

```python
import os
import sys
import shutil
import subprocess
import tempfile
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image

repo = Path("/content/SAM")

source = Path(
    "/content/stylegan2-ada-pytorch/generated_faces/seed0000.png"
)

models = [
    (
        "Before training",
        repo / "pretrained_models/sam_ffhq_aging.pt",
    ),
    (
        "After training",
        Path(
            "/content/sam_age_fixed_20261118_124839_460862/"
            "checkpoints/iteration_100.pt"
        ),
    ),
]

ages = [2, 10, 20, 40, 60, 80]

for path in [repo, source] + [model for _, model in models]:
    if not path.exists():
        raise FileNotFoundError(f"الملف مش موجود: {path}")

# مجلد جديد للتجربة؛ الملفات القديمة هتفضل زي ما هي
work = Path(tempfile.mkdtemp(prefix="sam_ages_", dir="/content"))
input_dir = work / "input"
input_dir.mkdir()
shutil.copy2(source, input_dir / "face.png")

env = os.environ.copy()
env["PYTHONPATH"] = str(repo) + os.pathsep + env.get("PYTHONPATH", "")
env["PYTHONUNBUFFERED"] = "1"

rows = []

for label, checkpoint in models:
    output_dir = work / (
        "before" if label == "Before training" else "after"
    )

    command = [
        sys.executable,
        "scripts/inference_side_by_side.py",
        f"--exp_dir={output_dir}",
        f"--checkpoint_path={checkpoint}",
        f"--data_path={input_dir}",
        "--test_batch_size=1",
        "--test_workers=0",
        "--target_age=" + ",".join(map(str, ages)),
    ]

    print(f"\nجاري الاختبار: {label}", flush=True)

    result = subprocess.run(
        command,
        cwd=str(repo),
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
    )

    if result.returncode != 0:
        print(result.stdout[-8000:])
        raise RuntimeError(
            f"اختبار {label} توقف. ابعتلي الخطأ اللي ظهر فوق."
        )

    output = output_dir / "inference_side_by_side" / "face.png"
    if not output.is_file():
        print(result.stdout[-4000:])
        raise FileNotFoundError(f"صورة النتيجة مش موجودة: {output}")

    with Image.open(output) as image:
        image = image.convert("RGB")
        width, height = image.size

        # SAM حفظ في تجربتنا السابقة: الأصل ثم صورة لكل عمر
        count = len(ages) + 1
        if width % count != 0:
            raise ValueError(
                f"أبعاد النتيجة غير متوقعة: {width}×{height}. "
                "ابعتلي الرسالة دي قبل ما نقسم الصور."
            )

        panel_width = width // count

        if abs(panel_width - height) > max(2, height * 0.05):
            raise ValueError(
                f"اللوحات مش بالمقاس المتوقع: {width}×{height}."
            )

        panels = [
            image.crop((
                i * panel_width, 0,
                (i + 1) * panel_width, height,
            ))
            for i in range(count)
        ]

    # أول لوحة هي الأصل، والباقي حسب الأعمار المطلوبة
    rows.append((label, panels[1:]))
    print("الاختبار خلص:", label)

# عرض الأصل منفصل، ثم صفين للأعمار
fig = plt.figure(figsize=(18, 8))
grid = fig.add_gridspec(3, len(ages), height_ratios=[0.9, 1, 1])

original_ax = fig.add_subplot(grid[0, 2:4])
with Image.open(source) as image:
    original_ax.imshow(image.convert("RGB"))
original_ax.set_title("Original input")
original_ax.axis("off")

for row_index, (label, panels) in enumerate(rows, start=1):
    for col, (age, panel) in enumerate(zip(ages, panels)):
        ax = fig.add_subplot(grid[row_index, col])
        ax.imshow(panel)
        ax.set_title(f"Target age: {age}")
        ax.axis("off")

    fig.text(
        0.01,
        0.59 if row_index == 1 else 0.24,
        label,
        rotation=90,
        va="center",
        fontsize=12,
    )

fig.suptitle(
    "Original input above | Ages ordered from youngest to oldest",
    fontsize=14,
)
plt.tight_layout(rect=[0.04, 0, 1, 0.95])
plt.show()

print("\nمجلد التجربة:", work)
```

```python
import os
import gc
import torch
import matplotlib.pyplot as plt

from PIL import Image, ImageOps, ImageDraw, ImageFilter
from diffusers import StableDiffusionInpaintPipeline

if not torch.cuda.is_available():
    raise RuntimeError("الـGPU مش شغال في Colab.")

if "filename" not in globals() or not os.path.isfile(filename):
    raise FileNotFoundError(
        "الصورة مش موجودة في جلسة Colab. "
        "ارفع نفس صورة المومياء في Colab باستخدام خلية الرفع السابقة."
    )

print("الصورة المستخدمة:", filename)

# توفير الذاكرة لو موديل التجربة السابقة موجود
if "pipe" in globals():
    del pipe

gc.collect()
torch.cuda.empty_cache()

# الحفاظ على النسب الأصلية داخل صورة 512×512
original = Image.open(filename).convert("RGB")
image = ImageOps.pad(original, (512, 512), color="black")

# منطقة الرأس تقريبية ومناسبة للصورة اللي بعتها.
# الأبيض = المنطقة اللي هنسمح للموديل يغيّرها
mask = Image.new("L", original.size, 0)
draw = ImageDraw.Draw(mask)

w, h = original.size
draw.ellipse(
    (int(w * 0.13), int(h * 0.05),
     int(w * 0.88), int(h * 0.73)),
    fill=255
)

# نطبّق نفس التحجيم المستخدم للصورة
mask = ImageOps.pad(mask, (512, 512), color=0)
mask = mask.filter(ImageFilter.GaussianBlur(8))

print("جاري تحميل نموذج تعديل المنطقة...")

inpaint = StableDiffusionInpaintPipeline.from_pretrained(
    "stable-diffusion-v1-5/stable-diffusion-inpainting",
    torch_dtype=torch.float16,
).to("cuda")

inpaint.enable_attention_slicing()

prompt = (
    "A realistic portrait photograph of a living adult human, "
    "natural healthy skin, human face, open eyes, natural nose "
    "and lips, looking toward the camera. Preserve the approximate "
    "head position and facial proportions of the source image."
)

negative_prompt = (
    "mummy, skeleton, skull, dried skin, exposed bones, "
    "anatomical illustration, wrappings, horror, distorted face, "
    "extra eyes, deformed facial features"
)

print("جاري إنشاء تصور احتمالي...")

result = inpaint(
    prompt=prompt,
    negative_prompt=negative_prompt,
    image=image,
    mask_image=mask,
    strength=0.9,
    guidance_scale=8.0,
    num_inference_steps=35,
    generator=torch.Generator("cuda").manual_seed(42),
).images[0]

fig, axes = plt.subplots(1, 3, figsize=(15, 6))

axes[0].imshow(image)
axes[0].set_title("Original")

axes[1].imshow(mask, cmap="gray")
axes[1].set_title("Area to replace")

axes[2].imshow(result)
axes[2].set_title("Hypothetical portrait")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
import torch
import numpy as np
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
from diffusers import StableDiffusionControlNetPipeline, ControlNetModel
from transformers import DPTImageProcessor, DPTForDepthEstimation
import glob

# 1. التأكد من عمل الـ GPU
if not torch.cuda.is_available():
    raise RuntimeError("الـ GPU مش شغّال. فعّله من Runtime > Change runtime type.")

# 2. جلب صورة المومياء المرفقة
uploaded_files = glob.glob("/content/*.jpg") + glob.glob("/content/*.png") + glob.glob("*.jpg")
img_path = uploaded_files[0] if uploaded_files else "images.jpg"

original_image = Image.open(img_path).convert("RGB")
original_image = ImageOps.fit(original_image, (512, 512))

print("جاري تحميل ControlNet واستخراج خريطة العمق الدقيقة للمومياء...")

# تحميل ControlNet الخاص بالعمق
controlnet = ControlNetModel.from_pretrained(
    "lllyasviel/sd-controlnet-depth",
    torch_dtype=torch.float16
)

pipe = StableDiffusionControlNetPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    controlnet=controlnet,
    torch_dtype=torch.float16
)
pipe = pipe.to("cuda")
pipe.enable_attention_slicing()

# استخراج خريطة العمق بدقة عالية (DPT Model)
depth_processor = DPTImageProcessor.from_pretrained("Intel/dpt-large")
depth_model = DPTForDepthEstimation.from_pretrained("Intel/dpt-large").to("cuda")

inputs = depth_processor(images=original_image, return_tensors="pt").to("cuda")
with torch.no_grad():
    depth_image = depth_model(**inputs).predicted_depth

depth_image = depth_image.squeeze().cpu().float().numpy()
depth_image = (depth_image * 255 / depth_image.max()).astype(np.uint8)
depth_image = Image.fromarray(depth_image).convert("RGB")
depth_image = ImageOps.fit(depth_image, (512, 512))

print("جاري التوليد بدقة الحفاظ على العمر الحقيقي (رجل مسسن حي)...")

# Prompt دقيق يطلب نفس العمر المتقدم للمومياء (في الثمانينات) ولكن بجلد بشري حي وطبيعي
prompt = (
    "A photorealistic portrait of an ancient historical elderly man in his 80s, "
    "alive human skin texture, realistic wrinkles, natural aged face, "
    "preserving exact bone structure and facial geometry of the mummy, "
    "professional museum photography, masterpiece, 8k resolution."
)

# Negative Prompt لمنع الشباب، التماثيل، المومياوات، والجفاف
negative_prompt = (
    "young, teenager, in his 20s, in his 30s, mummy, skeleton, dead, decay, "
    "golden mask, statue, painting, illustration, deformed, ugly, bad anatomy"
)

generator = torch.Generator("cuda").manual_seed(42)

# توليد الوجه الحي مع زيادة قوة مطابقة الهيكل (conditioning scale = 0.85)
result_image = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    image=depth_image,
    controlnet_conditioning_scale=0.85, # التزام صارم بهيكل الجمجمة والوجه الأصلي
    guidance_scale=7.5,
    num_inference_steps=45,
    generator=generator
).images[0]

# عرض النتائج بشكل احترافي
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(original_image)
axes[0].set_title("Original Mummy")
axes[0].axis("off")

axes[1].imshow(depth_image)
axes[1].set_title("Bone Depth Map")
axes[1].axis("off")

axes[2].imshow(result_image)
axes[2].set_title("Living Elderly Man (Same True Age)")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
